```python
# Basic Single Feature Approach


x = [ 1, 2, 3, 4, 5]
y = [ 2, 4, 6, 8, 10]

w = 0
b = 0
i = 0
diffsum = 0
lr = 0.06


for i in range(100):
    i = 0
    wg = 0
    bg = 0
    diffsum = 0
    for xf in x:
        pred = w * xf + b
        diffsum += (pred - y[i]) ** 2
        wg += (pred - y[i]) * xf
        bg += (pred - y[i])
        i += 1

    mse = (1 / len(x)) * diffsum
    dw = (2 / len(x)) * wg
    db = (2 / len(x)) * bg
    w = w - lr * dw
    b = b - lr * db


print(w, b)
```

```python
# Multi Feature Approach

x = [
    [1, 2],
    [2, 1],
    [3, 2],
    [4, 3],
    [5, 1]
]

y = [13, 15, 23, 31, 30]

w = 0
b = 0
i = 0
diffsum = 0
lr = 0.01

weights = []
wgacc = []

for present in x[0]:
    weights.append(0)
    wgacc.append(0)


for i in range(10000):
    i = 0
    bgacc = 0
    diffsumm = 0
    wgacc = [0] * len(x[0])
    iter = (len(x[0]))
    for xf in x:
        predacc = 0
        for g in range(iter):
            predacc += (weights[g] * xf[g])

        predm = predacc + b
        
        for g in range(iter):
            wgacc[g] += ( predm - y[i]) * xf[g]

        bgacc += (predm - y[i])
        diffsumm += (predm - y[i]) ** 2
        i += 1


    mse = (1 / len(x)) * diffsumm
    db = (2 / len(x)) * bgacc
    b = b - lr * db

    for i in range(len(wgacc)):
        dw = (2 / len(x)) * wgacc[i]
        weights[i] = weights[i] - lr * dw


print("Weights:", weights)
print("Bias:", b)
print("MSE:", mse)
```
